```python
import pandas as pd

df = pd.read_csv('heart_disease_uci.csv')

df.head()
```

```python
display(df.head())
print("Dataset Shape:", df.shape)
df.info() #inf of columns and data types
```

```python
print(df.isnull().sum())
```

```python
numeric_cols_with_na = ['trestbps', 'chol', 'thalch', 'oldpeak']

for col in numeric_cols_with_na:
  median_val = df[col].median()
  df[col].fillna(median_val, inplace = True)

  print(df[numeric_cols_with_na].isnull().sum())
```

```python
categorical_cols = ['fbs', 'restecg', 'exang', 'slope', 'ca', 'thal']
print(df[categorical_cols].isnull().sum())
```

```python
for col in categorical_cols:
  mode_val = df[col].mode()[0]
  df[col].fillna(mode_val, inplace = True)

print("Total Missing Values in DataFrame:\n", df.isnull().sum())
```

```python
object_cols = df.select_dtypes(include=['object']).columns
print("categorical columns to encoding:", list(object_cols))
```

```python
df = pd.get_dummies(df, columns=['sex', 'dataset', 'cp', 'restecg', 'slope', 'thal'], drop_first = True)
display(df.head(3))

print("new dataset shape:", df.shape)
```

```python
import numpy as np

numeric_cols = ['age', 'trestbps', 'chol', 'thalch', 'oldpeak']
print("shape before IQR:", df.shape)

for col in numeric_cols:
  Q1 = df[col].quantile(.25)
  Q3 = df[col].quantile(.75)

IQR = Q3 - Q1
lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

df = df[(df[col] >= lower_bound) & (df[col] <= upper_bound)]

print("shape after IQR:", df.shape)
```

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()
df[numeric_cols] = scaler.fit_transform(df[numeric_cols])

display(df.head(5))

print("final dataset after EDA:", df.shape)
```

Exploratory Data Analysis

```python
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

sex_col = [col for col in df.columns if 'sex_' in col]
if sex_col:
    sns.countplot(data=df, x=sex_col[0], hue=(df['num'] > 0), ax=axes[0], palette='Set1')
    axes[0].set_title('Diagnosis Rate by Gender')
    axes[0].set_xlabel('Gender (Encoded)')
else:
    sns.countplot(data=df, x='sex', hue=(df['num'] > 0), ax=axes[0], palette='Set1')

cp_cols = [col for col in df.columns if 'cp_' in col]
if cp_cols:
    cp_df = df[cp_cols].idxmax(axis=1).value_counts().reset_index()
    sns.barplot(data=df, x='cp_non-anginal', hue=(df['num'] > 0), ax=axes[1], palette='Set2')
    axes[1].set_title('Diagnosis Rate by Chest Pain Type')
else:
    sns.countplot(data=df, x='cp', hue=(df['num'] > 0), ax=axes[1], palette='Set2')

plt.tight_layout()
plt.show()
```

"ML model"🤩

```python
from sklearn.model_selection import train_test_split

x = df.drop(columns=['num'])
y = (df['num'] > 0).astype(int)

x_train, x_test, y_train, y_test = train_test_split(x, y, test_size = .2, random_state = 42, stratify=y)

print("Training set shape (x_train):", x_train.shape)
print("Testing set shape (x_test):", x_test.shape)
```

```python
from sklearn.ensemble import RandomForestClassifier

rf_model = RandomForestClassifier(n_estimators = 100, random_state = 42)

rf_model.fit(x_train, y_train)

print("model succeeded")
```

```python
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix

y_pred = rf_model.predict(x_test)

accuracy = accuracy_score(y_test, y_pred)
print(f"Model Accuracy: {accuracy * 100:.2f}%\n")

print("Classification Report:")
print(classification_report(y_test, y_pred))
```

```python
import gradio as gr
import pandas as pd

features_to_use = [col for col in x.columns if col != 'id']

def predict_heart_disease(*args):
    input_data = dict(zip(features_to_use, args))
    input_data['id'] = 1
    input_df = pd.DataFrame([input_data])[x.columns]

    prediction = lgb_model.predict(input_df)[0]
    probability = lgb_model.predict_proba(input_df)[0][1]

    if prediction == 1:
        return f"Warning: High risk of heart disease (Probability: {probability*100:.1f}%)"
    else:
        return f"Safe: Low risk of heart disease (Probability: {probability*100:.1f}%)"

defaults = {
    'age': 50, 'sex': 'Male', 'dataset': 'Cleveland', 'cp': 'typical angina',
    'trestbps': 120, 'chol': 200, 'fbs': False, 'restecg': 'normal',
    'thalch': 150, 'exang': False, 'oldpeak': 1.0, 'slope': 'flat',
    'ca': 0.0, 'thal': 'normal'
}

inputs = [gr.Number(label=col, value=defaults.get(col, 0)) for col in features_to_use]

demo = gr.Interface(
    fn=predict_heart_disease,
    inputs=inputs,
    outputs="text",
    title="Heart Disease Prediction App",
    description="Enter medical details below to predict heart disease risk using LightGBM."
)

demo.launch(share=True)
```
